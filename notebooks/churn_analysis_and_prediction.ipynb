```python
# Helper function to display missing data by feature
# https://stackoverflow.com/questions/26266362/how-to-count-the-nan-values-in-a-column-in-pandas-dataframe#26266451
def missing_values_table(df):
    # Calculate the number of missing values per column
    mis_val = df.isnull().sum()

    # Calculate the percentage of missing values per column
    mis_val_percent = 100 * df.isnull().sum() / len(df)

    # Create a table combining the missing values and their percentages
    mis_val_table = pd.concat([mis_val, mis_val_percent], axis=1)

    # Rename the columns of the table
    mis_val_table_ren_columns = mis_val_table.rename(columns={0: 'No. Missing Values', 1: '%'})

    # Filter out columns with no missing values, sort in descending order by percentage, and round to 1 decimal place
    mis_val_table_ren_columns = mis_val_table_ren_columns[mis_val_table_ren_columns.iloc[:, 1] != 0].sort_values('%', ascending=False).round(1)

    # Print the number of features with missing values
    print(f"There are {mis_val_table_ren_columns.shape[0]} features with missing values.")

    return mis_val_table_ren_columns

# Call the function to display the missing values table for the DataFrame 'df'
missing_values_table(df)
```

```python
#| code-fold: true

import numpy as np
import plotly.graph_objs as go
import plotly.express as px
import scipy.stats as stats

# Fit a normal distribution to the 'days_since_last_login' data
mu, sigma = stats.norm.fit(df['days_since_last_login'])

# Create a histogram of the 'days_since_last_login' data
hist_data = go.Histogram(
    x=df['days_since_last_login'],
    nbinsx=50,
    name="Histogram",
    opacity=0.75,
    histnorm='probability density',
    marker=dict(color='#5abbf9')
)

# Generate data for a fitted normal distribution
x_norm = np.linspace(df['days_since_last_login'].min(), df['days_since_last_login'].max(), 100)
y_norm = stats.norm.pdf(x_norm, mu, sigma)

# Create a scatter plot for the fitted normal distribution
norm_data = go.Scatter(
    x=x_norm,
    y=y_norm,
    mode="lines",
    name=f"Normal dist. (μ={mu:.2f}, σ={sigma:.2f})",
    line=dict(color="#d8d8d8")
)

# Create a figure that combines the histogram and normal distribution plot
fig = go.Figure(data=[hist_data, norm_data])

# Configure the layout of the figure
fig.update_layout(
    title="Distribution of Days Since Last Login",
    xaxis_title="Days Since Last Login",
    yaxis_title="Probability Density",
    legend_title_text="Fitted Normal Distribution",
    plot_bgcolor='rgba(32, 32, 32, 1)',
    paper_bgcolor='rgba(32, 32, 32, 1)',
    font=dict(color='white')
)

# Create a probability-probability (Q-Q) plot
qq_data = stats.probplot(df['days_since_last_login'], dist="norm")
qq_fig = px.scatter(
    x=qq_data[0][0],
    y=qq_data[0][1],
    labels={'x': 'Theoretical Quantiles', 'y': 'Ordered Values'},
    color_discrete_sequence=["#5abbf9"]
)

# Configure the layout of the Q-Q plot
qq_fig.update_layout(
    title="Q-Q Plot",
    plot_bgcolor='rgba(32, 32, 32, 1)',
    paper_bgcolor='rgba(32, 32, 32, 1)',
    font=dict(color='white')
)

# Fit a linear regression line to the Q-Q plot
slope, intercept, r_value, p_value, std_err = stats.linregress(qq_data[0][0], qq_data[0][1])
line_x = np.array(qq_data[0][0])
line_y = intercept + slope * line_x

# Add the linear regression line to the Q-Q plot
line_data = go.Scatter(
    x=line_x,
    y=line_y,
    mode="lines",
    name="Normal Line",
    line=dict(color="#d8d8d8")
)

qq_fig.add_trace(line_data)

# Update figure axes properties
fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=True, gridcolor='gray')

# Display the figures
fig.show()
qq_fig.show()
```

```python
def find_outlier_bounds(df, num_cols, multiplier=1.5):
    """
    Find the lower and upper bounds for identifying outliers in numerical columns using the IQR method.

    Parameters:
        df (DataFrame): The DataFrame containing the data.
        num_cols (list): A list of column names containing numerical data.
        multiplier (float, optional): Multiplier for the IQR. Default is 1.5.

    Returns:
        outlier_bounds (dict): A dictionary containing lower and upper bounds for each numerical column.
    """
    outlier_bounds = {}

    for col_name in keyvars_df:
        thirdq, firstq = df[col_name].quantile(0.75), df[col_name].quantile(0.25)
        interquartile_range = multiplier * (thirdq - firstq)
        upper_bound = interquartile_range + thirdq
        lower_bound = firstq - interquartile_range
        outlier_bounds[col_name] = (lower_bound, upper_bound)

    return outlier_bounds
```

```python
keyvars_df = ['days_since_last_login', 'avg_time_spent', 'avg_transaction_value',
           'points_in_wallet', 'avg_frequency_login_days','age']
outlier_bounds = find_outlier_bounds(df, keyvars_df)
for col_name, (lower_bound, upper_bound) in outlier_bounds.items():
    print(f"Column: {col_name}, Lower Bound: {lower_bound}, Upper Bound: {upper_bound}")
```

```python
from sklearn.base import BaseEstimator, TransformerMixin
class OutlierTrans(BaseEstimator, TransformerMixin):
    def __init__(self, threshold=1.5):
        self.threshold = threshold

    def fit(self, X, y=None):
        # This transformer does not need any fitting, so we return self.
        return self

    def transform(self, X, y=None):
        X_new = X.copy()
        for col in X_new.columns:
            # Calculate the 1st and 3rd quartiles for the current column
            third_q, first_q = np.percentile(X_new[col], [75, 25])
            # Calculate the interquartile range
            interquartile_range = self.threshold * (third_q - first_q)
            # Define the outlier bounds
            outlier_high, outlier_low = third_q + interquartile_range, first_q - interquartile_range
            # Replace values outside the outlier bounds with NaN
            X_new.loc[(X_new[col] > outlier_high) | (X_new[col] < outlier_low), col] = np.nan
        return X_new.values
```

```python
class MakeOrdinal(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None):
        # This transformer does not need any fitting, so we return self.
        return self

    def transform(self, X, y=None):
        X_new = X.copy()
        for col in X_new.columns:
            # Get unique categories and sort them
            cats = np.sort(X_new[col].unique())
            # Replace categorical values with their ordinal values
            X_new[col] = X_new.apply(lambda x: int(np.where(cats == x[col])[0]), axis=1)
        return X_new.values
```

```python
class ReplaceVals(BaseEstimator, TransformerMixin):
    def __init__(self, repdict):
        self.repdict = repdict

    def fit(self, X, y=None):
        # This transformer does not need any fitting, so we return self.
        return self

    def transform(self, X, y=None):
        X_new = X.copy().replace(self.repdict)
        return X_new.values
```

```python
# Numerical columns for analysis
num_cols = ['days_since_last_login', 'avg_time_spent', 'avg_transaction_value',
            'points_in_wallet', 'avg_frequency_login_days']

# Special columns that may require special handling
special_col1 = ['age']
special_col2 = ['complaint_status']
special_col3 = ['membership_category']

# Categorical columns for analysis
cat_cols = ['gender', 'region_category', 'preferred_offer_types',
            'medium_of_operation', 'internet_option', 'feedback']

# Binary columns (having two categories) for analysis
binary_cols = ['joined_through_referral', 'used_special_discount',
               'offer_application_preference', 'past_complaint']

# Target column for analysis
target_col = ['churn_risk_score']

# Columns not in use for the analysis
column_not_in_use = ['security_no', 'referral_id', 'joining_date', 'last_visit_time', 'Unnamed: 0']
```

```python
from sklearn.model_selection import train_test_split

# Define the feature columns
feature_cols = num_cols + binary_cols + cat_cols + special_col1 + special_col2

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    df[feature_cols],
    df[['churn_risk_score']],
    test_size=0.2,
    random_state=0
)
```

```python
ohe = OneHotEncoder(drop='first', sparse=False)
standtrans = make_pipeline(
    OutlierTrans(3),
    SimpleImputer(strategy="median"),
    StandardScaler()
)
spectrans1 = make_pipeline(
    MakeOrdinal(),
    StandardScaler()
)
spectrans2 = make_pipeline(
    ReplaceVals(rep_dict),
    StandardScaler()
)
spectrans3 = make_pipeline(
    ReplaceVals(rep_dict)
)
bintrans = make_pipeline(ohe)
cattrans = make_pipeline(ohe)

coltrans = ColumnTransformer(
    transformers=[
        ("stand", standtrans, num_cols),
        ("spec1", spectrans1, special_col1),
        ("spec2", spectrans2, special_col2),
        ("bin", bintrans, binary_cols),
        ("cat", cattrans, cat_cols),
    ]
)
```

```python
# Create a DecisionTreeClassifier
dtc_example = DecisionTreeClassifier(min_samples_leaf=5, max_depth=2)

pipe0 = make_pipeline(coltrans, dtc_example)

# Fit the pipeline to your training data
pipe0.fit(X_train, y_train.values.ravel())
```

```python
# Get binary column names from the pipeline
new_binary_cols = pipe0.named_steps['columntransformer'] \
    .named_transformers_['bin'] \
    .named_steps['onehotencoder'] \
    .get_feature_names_out(input_features=binary_cols)

# Get categorical column names from the pipeline
new_cat_cols = pipe0.named_steps['columntransformer'] \
    .named_transformers_['cat'] \
    .named_steps['onehotencoder'] \
    .get_feature_names_out(input_features=cat_cols)

# Combine all column names into a single array
new_cols = np.concatenate((
    np.array(num_cols + special_col1 + special_col2),
    new_binary_cols,
    new_cat_cols
))
```

```python
# Get feature importances from the trained DecisionTreeClassifier
feature_imp = pipe0.named_steps['decisiontreeclassifier'].tree_.compute_feature_importances(normalize=False)

# Create a mask to filter features with importances greater than 0
feature_impgt0 = feature_imp > 0

# Stack feature importances with corresponding feature names
feature_implabs = np.column_stack((feature_imp.ravel(), new_cols))
```

```python
pred = pipe0.predict(X_test)
accuracy = skmet.accuracy_score(y_test.values.ravel(), pred)
sensitivity = skmet.recall_score(y_test.values.ravel(), pred)
specificity = skmet.recall_score(y_test.values.ravel(), pred, pos_label=0)
precision = skmet.precision_score(y_test.values.ravel(), pred)

print("accuracy: %.2f, sensitivity: %.2f, specificity: %.2f, precision: %.2f" % (accuracy, sensitivity, specificity, precision))
```
